```python
def clean_na(dataset):
    columns_check_na = ['Price_CLP', 'Comuna', 'Dorms', 'Baths', 'Built Area', 'Total Area']
    # columns_check_na = ['Price_CLP', 'Comuna']
    data_clean_na = dataset.dropna(subset=columns_check_na)
    data_clean_na.loc[:, 'Parking'] = data_clean_na['Parking'].fillna(0.0)
    # data_clean_na.loc[:, 'Realtor'] = data_clean_na['Realtor'].fillna('Desconocido')
    # data_clean_na['Dorms'] = data_clean_na['Dorms'].astype(int)
    # data_clean_na['Baths'] = data_clean_na['Baths'].astype(int)
    # data_clean_na['Parking'] = data_clean_na['Parking'].astype(int)
    return data_clean_na

def limit_values(dataset, column, max, min):
    return dataset.drop(dataset[(dataset[column] >= max) | (dataset[column] <= min)].index)

def add_lat_lon_by_comn(dataset, column = 'Comuna'):
    dataset['coords'] = dataset[column].map(comuna_coords) #Añadir coords
    dataset[['latitude', 'longitude']] = dataset['coords'].apply(
        lambda x: pd.Series(x) if pd.notna(x) else pd.Series({'lat': None, 'lon': None})
    ) 
    return dataset.drop(columns=['coords'])
```

preprocesando dataset de casas usadas chile RM Julio 2023

-  Eliminando columnas con valores importantes desconocidos
-  Supuesto de que el valor Parking = NaN significa que no tiene estacionamiento (= 0)
-  En caso de no tener corredor mencionado (col. Realtor) se marca como desconocido

```python
data_jul_23 = pd.read_csv('../data/raw/kaggle - Casas Chile RM jul-23.csv')

print("Tamaño original: " + str(data_jul_23.shape))
data_clean_jul_23 = clean_na(dataset=data_jul_23)
print("Despues de limpiar los datos: " + str(data_clean_jul_23.shape))
data_clean_jul_23 = limit_values(data_clean_jul_23, 'Total Area', 150, 15)
print("Despues de limitar el area: " + str(data_clean_jul_23.shape))
data_clean_jul_23 = add_lat_lon_by_comn(data_clean_jul_23)

data_preproccesed = data_clean_jul_23.copy()

data_preproccesed = data_clean_jul_23.rename(columns={
    'Price_CLP': 'Price',
    'Dorms': 'Room',
    'Baths': 'Bathroom',
    'latitude': 'Latitude',
    'longitude': 'Longitude',
    'Built Area': 'BuildingArea',
    'Total Area': 'TotalArea'
    })
columns_drop = ['Comuna', 'Ubicacion', 'id', 'Realtor', 'Price_UF', 'Price_USD']

data_preproccesed = data_preproccesed.drop(columns=columns_drop)
data_preproccesed.to_csv('../data/processed/kaggle - Casas Chile RM jul-23 - proccessed.csv', index=False)
print('Dataset preprocesado y guarado.')
```
